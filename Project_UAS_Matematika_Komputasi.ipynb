# Metode Bisection

Metode Bisection digunakan untuk mencari akar suatu persamaan dengan cara membagi interval menjadi dua bagian secara berulang.

Metode ini memerlukan batas bawah dan batas atas yang memenuhi syarat bahwa nilai fungsi pada kedua batas tersebut memiliki tanda yang berbeda.

Rumus yang digunakan pada Metode Bisection adalah:

xt = (xa + xb) / 2

Keterangan:

xt = titik tengah interval

xa = batas bawah

xb = batas atas


```python
# Import Library

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Penjelasan Library

NumPy digunakan untuk membantu perhitungan numerik.

Pandas digunakan untuk membuat tabel hasil iterasi agar lebih mudah dibaca.

Matplotlib digunakan untuk membuat grafik fungsi sehingga akar persamaan dapat dilihat secara visual.

```python
# Input Data dari Pengguna
fungsi = input("Masukkan fungsi yang ingin dicari akarnya: ")

batas_bawah = float(input("Masukkan batas bawah interval: "))

batas_atas = float(input("Masukkan batas atas interval: "))

toleransi = float(input("Masukkan tingkat ketelitian yang diinginkan: "))
```

Penjelasan Input

Fungsi adalah persamaan yang akan dicari akarnya.

Batas bawah dan batas atas digunakan sebagai interval awal pencarian akar.

Tingkat ketelitian (toleransi) digunakan untuk menentukan kapan proses iterasi dihentikan.
Semakin kecil nilainya, semakin teliti hasil yang diperoleh.

```python
# Membuat Fungsi Matematika
def hitung_fungsi(x):
    return eval(fungsi)
```

## Penjelasan Fungsi

Fungsi ini digunakan untuk menghitung nilai fungsi yang dimasukkan pengguna.

Sebagai contoh, jika fungsi yang dimasukkan adalah x**3 - 4*x - 9, maka program dapat menghitung nilai fungsi tersebut pada setiap nilai x yang diperlukan.

```python
# Memeriksa Syarat Metode Bisection
if hitung_fungsi(batas_bawah) * hitung_fungsi(batas_atas) > 0:
    print("Interval yang dimasukkan tidak memenuhi syarat Metode Bisection")
else:
    print("Interval memenuhi syarat")
```

## Validasi Interval

Metode Bisection mensyaratkan bahwa fungsi pada batas bawah dan batas atas memiliki tanda yang berbeda.

Jika syarat ini tidak terpenuhi, maka proses pencarian akar tidak dapat dilakukan.

```python
# Perhitungan Metode Bisection

data_iterasi = []

jumlah_iterasi = 0

while abs(batas_atas - batas_bawah) > toleransi:

    titik_tengah = (batas_bawah + batas_atas) / 2

    data_iterasi.append([
        jumlah_iterasi + 1,
        batas_bawah,
        batas_atas,
        titik_tengah
    ])

    if hitung_fungsi(batas_bawah) * hitung_fungsi(titik_tengah) < 0:
        batas_atas = titik_tengah
    else:
        batas_bawah = titik_tengah

    jumlah_iterasi += 1

akar = (batas_bawah + batas_atas) / 2

print("Akar persamaan =", akar)

print("Jumlah iterasi =", jumlah_iterasi)
```

## Melakukan Perhitungan Metode Bisection

Pada bagian ini program melakukan proses pencarian akar secara berulang sampai diperoleh hasil yang memenuhi tingkat ketelitian yang diinginkan.

```python
# Menampilkan Tabel Iterasi

tabel_iterasi = pd.DataFrame(
    data_iterasi,
    columns=[
        "Iterasi",
        "Batas Bawah",
        "Batas Atas",
        "Titik Tengah"
    ]
)

tabel_iterasi
```

# Metode Newton-Raphson

Metode Newton-Raphson digunakan untuk mencari akar suatu persamaan menggunakan pendekatan garis singgung.

Metode ini memerlukan fungsi dan turunan fungsi untuk memperoleh pendekatan akar yang baru pada setiap iterasi.

Rumus yang digunakan pada Metode Newton-Raphson adalah:

x_(n+1) = x_n - f(x_n) / f'(x_n)

Keterangan:

x_(n+1) = pendekatan akar yang baru

x_n = pendekatan akar sebelumnya

f(x_n) = nilai fungsi pada x_n

f'(x_n) = nilai turunan fungsi pada x_n

```python
# Import Library Tambahan

import sympy as sp
```

Library SymPy digunakan untuk menghitung turunan fungsi secara otomatis sehingga pengguna tidak perlu mencari turunannya secara manual.

```python
# Input Data dari Pengguna

fungsi = input("Masukkan fungsi yang ingin dicari akarnya: ")

tebakan_awal = float(input("Masukkan tebakan awal: "))

toleransi = float(input("Masukkan tingkat ketelitian yang diinginkan: "))
```

# Penjelasan Input Data

Pada bagian ini pengguna diminta memasukkan fungsi yang akan dicari akarnya, nilai tebakan awal, dan nilai toleransi yang digunakan sebagai batas ketelitian hasil.

```python
# Membuat Fungsi dan Turunan Fungsi

x = sp.Symbol("x")

fungsi_sympy = sp.sympify(fungsi)

turunan_fungsi = sp.diff(fungsi_sympy, x)

print("Turunan fungsi =", turunan_fungsi)
```

# Penjelasan Input
Pada bagian ini program mengubah fungsi yang dimasukkan pengguna menjadi bentuk matematika yang dapat dihitung oleh Python.

Program juga menghitung turunan fungsi yang akan digunakan pada Metode Newton-Raphson.

```python
# Perhitungan Metode Newton-Raphson

data_iterasi = []

akar = tebakan_awal

jumlah_iterasi = 0

while True:

    nilai_fungsi = float(fungsi_sympy.subs(x, akar))

    nilai_turunan = float(turunan_fungsi.subs(x, akar))

    akar_baru = akar - (nilai_fungsi / nilai_turunan)

    data_iterasi.append([
        jumlah_iterasi + 1,
        akar,
        akar_baru
    ])

    if abs(akar_baru - akar) < toleransi:
        break

    akar = akar_baru

    jumlah_iterasi += 1

print("Akar persamaan =", akar_baru)

print("Jumlah iterasi =", jumlah_iterasi)
```

# Penjelasan Perhitungan

Pada bagian ini program melakukan proses pencarian akar persamaan menggunakan Metode Newton-Raphson. Nilai pendekatan akar akan terus diperbarui sampai hasil yang diperoleh memenuhi tingkat ketelitian yang telah ditentukan.

```python
# Menampilkan Tabel Iterasi

tabel_iterasi = pd.DataFrame(
    data_iterasi,
    columns=[
        "Iterasi",
        "x lama",
        "x baru"
    ]
)

tabel_iterasi
```

# Penjelasan Tabel Iterasi

Tabel ini digunakan untuk melihat perkembangan nilai pendekatan akar pada setiap iterasi sampai diperoleh hasil akhir yang diinginkan.

# Metode Secant

Metode Secant digunakan untuk mencari akar suatu persamaan menggunakan dua tebakan awal. Metode ini tidak memerlukan turunan fungsi sehingga lebih sederhana dibandingkan Metode Newton-Raphson.

Rumus yang digunakan pada Metode Secant adalah:

x_(n+1) = x_n - [f(x_n)(x_n - x_(n-1))] / [f(x_n) - f(x_(n-1))]

Keterangan:

x_(n+1) = nilai pendekatan akar yang baru

x_n = nilai pendekatan akar saat ini

x_(n-1) = nilai pendekatan akar sebelumnya

f(x_n) = nilai fungsi pada x_n

f(x_(n-1)) = nilai fungsi pada x_(n-1)


```python
\# Input Data dari Pengguna

fungsi = input("Masukkan fungsi yang ingin dicari akarnya: ")

tebakan_pertama = float(input("Masukkan tebakan pertama: "))

tebakan_kedua = float(input("Masukkan tebakan kedua: "))

toleransi = float(input("Masukkan tingkat ketelitian yang diinginkan: "))
```

# Penjelasan Input Data

Pada bagian ini pengguna diminta memasukkan fungsi yang akan dicari akarnya, dua tebakan awal, dan nilai toleransi sebagai batas ketelitian hasil yang diinginkan.

```python
# Membuat Fungsi Matematika
def hitung_fungsi(x):
    return eval(fungsi)
```

# Penjelasan Fungsi Matematika

Bagian ini digunakan agar fungsi yang dimasukkan pengguna dapat dibaca dan dihitung oleh program. Fungsi tersebut nantinya akan digunakan pada setiap proses iterasi.

```python
# Perhitungan Metode Secant
data_iterasi = []

jumlah_iterasi = 0

while True:

    nilai_x0 = hitung_fungsi(tebakan_pertama)

    nilai_x1 = hitung_fungsi(tebakan_kedua)

    akar_baru = tebakan_kedua - (
        nilai_x1 * (tebakan_kedua - tebakan_pertama)
    ) / (nilai_x1 - nilai_x0)

    data_iterasi.append([
        jumlah_iterasi + 1,
        tebakan_pertama,
        tebakan_kedua,
        akar_baru
    ])

    if abs(akar_baru - tebakan_kedua) < toleransi:
        break

    tebakan_pertama = tebakan_kedua
    tebakan_kedua = akar_baru

    jumlah_iterasi += 1

print("Akar persamaan =", akar_baru)

print("Jumlah iterasi =", jumlah_iterasi)
```

# Penjelasan Perhitungan

Pada bagian ini program melakukan perhitungan Metode Secant secara berulang. Iterasi akan berhenti ketika hasil yang diperoleh sudah memenuhi nilai toleransi yang ditentukan.

```python
# Menampilkan Tabel Iterasi

tabel_iterasi = pd.DataFrame(
    data_iterasi,
    columns=[
        "Iterasi",
        "Tebakan Pertama",
        "Tebakan Kedua",
        "Akar Baru"
    ]
)

tabel_iterasi
```

# Penjelasan Grafik Fungsi

Grafik ditampilkan untuk membantu melihat bentuk fungsi yang digunakan dan memperlihatkan posisi akar persamaan yang diperoleh.
